````python
from typing import TypedDict, Annotated, List
import operator

class AgentState(TypedDict):
    chat_history: List[str]
    user_input: str
    tool_output: str
````

````python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Initialize the OpenAI LLM
# It will use the OPENAI_API_KEY from environment variables
llm = ChatOpenAI(model="gpt-4o")

# Create a simple prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("user", "{input}")
])

# Create a chain to combine the prompt and LLM
chain = prompt | llm

print("LangChain chain with OpenAI LLM created.")
print("You can now invoke this 'chain' with an input.")
````

````python
from langchain_core.messages import BaseMessage, HumanMessage

# Define a simple agent node (a dummy function for now)
def agent_node(state):
    print("---AGENT NODE---")
    chat_history = state["chat_history"]
    user_input = state["user_input"]

    # In a real scenario, an LLM would process user_input and chat_history
    # to decide on an action or generate a response.

    if "tool" in user_input.lower():
        response = "Calling a tool..."
    else:
        response = f"Agent received: {user_input}. Current history: {chat_history}"

    return {"chat_history": chat_history + [response]}

# Define a simple tool node (a dummy function for now)
def tool_node(state):
    print("---TOOL NODE---")
    user_input = state["user_input"]

    # In a real scenario, this would execute a tool
    tool_result = f"Tool executed for: {user_input}"

    return {"tool_output": tool_result, "chat_history": state["chat_history"] + [tool_result]}
````

````python
# Example: Invoke the chain with an input
example_input = "What is the capital of France?"
response = chain.invoke({"input": example_input})

print(f"Question: {example_input}")
print(f"Answer: {response.content}")
````

````python
from langgraph.graph import StateGraph, END

# Define a conditional edge to decide whether to call a tool or end
def should_continue(state):
    if "tool" in state["user_input"].lower() and not state.get("tool_output"):
        return "tool"
    else:
        return "end"

# Build the graph
workflow = StateGraph(AgentState)

# Add nodes
workflow.add_node("agent", agent_node)
workflow.add_node("tool", tool_node)

# Set the entry point
workflow.set_entry_point("agent")

# Define edges
workflow.add_conditional_edges(
    "agent", # From agent
    should_continue, # The function to determine next step
    {
        "tool": "tool", # If should_continue returns "tool", go to tool_node
        "end": END # If should_continue returns "end", terminate
    }
)
workflow.add_edge("tool", END) # After tool execution, always terminate for this simple example

# Compile the graph
app = workflow.compile()
````

````python
# Example Usage
print("--- Agent activated with 'hello' ---")
inputs = {"user_input": "hello", "chat_history": [], "tool_output": ""}
result = app.invoke(inputs)
print(f"Final State: {result}")

print("\n--- Agent activated with 'use tool' ---")
inputs = {"user_input": "use tool", "chat_history": [], "tool_output": ""}
result = app.invoke(inputs)
print(f"Final State: {result}")
````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from transformers import AutoTokenizer
from langchain_core.documents import Document as LangchainDocument # Corrected import path
from typing import List, Optional # Added List and Optional imports
import pandas as pd # Ensure pandas is imported
import matplotlib.pyplot as plt # Ensure matplotlib is imported
from tqdm.notebook import tqdm # Ensure tqdm is imported

EMBEDDING_MODEL_NAME = "thenlper/gte-small"

# We use a hierarchical list of separators specifically tailored for splitting Markdown documents
# This list is taken from LangChain's MarkdownTextSplitter class
MARKDOWN_SEPARATORS = [
    "\n#{1,6} ",
    "```\n",
    "\n\\*\\*\\*+\n",
    "\n---+\n",
    "\n___+\n",
    "\n\n",
    "\n",
    " ",
    "",
]

def split_documents(
    chunk_size: int,
    knowledge_base: List[LangchainDocument],
    tokenizer_name: Optional[str] = EMBEDDING_MODEL_NAME,
) -> List[LangchainDocument]:
    """
    Split documents into chunks of maximum size `chunk_size` tokens and return a list of documents.
    """
    text_splitter = RecursiveCharacterTextSplitter.from_huggingface_tokenizer(
        AutoTokenizer.from_pretrained(tokenizer_name),
        chunk_size=chunk_size,
        chunk_overlap=int(chunk_size / 10),
        add_start_index=True,
        strip_whitespace=True,
        separators=MARKDOWN_SEPARATORS,
    )

    docs_processed = []
    for doc in knowledge_base:
        docs_processed += text_splitter.split_documents([doc])

    # Remove duplicates
    unique_texts = {}
    docs_processed_unique = []
    for doc in docs_processed:
        if doc.page_content not in unique_texts:
            unique_texts[doc.page_content] = True
            docs_processed_unique.append(doc)

    return docs_processed_unique


docs_processed = split_documents(
    512,  # We choose a chunk size adapted to our model
    RAW_KNOWLEDGE_BASE,
    tokenizer_name=EMBEDDING_MODEL_NAME,
)

# Let's visualize the chunk sizes we would have in tokens from a common model
tokenizer = AutoTokenizer.from_pretrained(EMBEDDING_MODEL_NAME)
lengths = [len(tokenizer.encode(doc.page_content)) for doc in tqdm(docs_processed)]
fig = pd.Series(lengths).hist()
plt.title("Distribution of document lengths in the knowledge base (in count of tokens)")
plt.show()
````

````python
from langchain.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores.utils import DistanceStrategy
import torch

embedding_model = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL_NAME,
    multi_process=True,
    model_kwargs={"device": "cuda" if torch.cuda.is_available() else "cpu"},
    encode_kwargs={"normalize_embeddings": True},  # Set `True` for cosine similarity
)

KNOWLEDGE_VECTOR_DATABASE = FAISS.from_documents(
    docs_processed, embedding_model, distance_strategy=DistanceStrategy.COSINE
)
````

````python
from transformers import pipeline
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

READER_MODEL_NAME = "HuggingFaceH4/zephyr-7b-beta"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
)
model = AutoModelForCausalLM.from_pretrained(
    READER_MODEL_NAME, quantization_config=bnb_config
)
tokenizer = AutoTokenizer.from_pretrained(READER_MODEL_NAME)

READER_LLM = pipeline(
    model=model,
    tokenizer=tokenizer,
    task="text-generation",
    do_sample=True,
    temperature=0.2,
    repetition_penalty=1.1,
    return_full_text=False,
    max_new_tokens=500,
)
````

````python
import json

def rag_answer_final(question: str, llm) -> str:
    print(f" Вопрос: '{question}'")
    query_embedding = embeddings.embed_query(question)
    results = client.query_points(collection_name="test", query=query_embedding, limit=5, with_payload=True)



    context_parts = []

    for i, point in enumerate(results.points[:5]):  # 5 лучших
        if point.payload and 'page_content' in point.payload:
            page_content = point.payload['page_content']

            try:
                data = json.loads(page_content)
                if 'chunk' in data:
                    text = data['chunk']
                else:
                    text = str(data)[:300]

            except (json.JSONDecodeError, TypeError):
                text = page_content[:300]

            if len(text) > 300:
                text = text[:297] + "..."

            context_parts.append(f"[Документ {i+1}] {text}")
            print(f"   Док {i+1}: {text[:80]}...")


        if context_parts:
            context = "\n\n".join(context_parts)
            prompt = f"""Ты - аналитик. Отвечай на вопросы ТОЛЬКО на основе предоставленной информации.
Если информации недостаточно для ответа, скажи: "В предоставленных документах нет точной информации по этому вопросу."

ИНФОРМАЦИЯ:
{context}

ВОПРОС: {question}

ОТВЕТ (на основе информации выше):"""
        else:
            prompt = f"Ответь на вопрос: {question}"


        if hasattr(llm, 'invoke'):
            response = llm.invoke(prompt)
            answer = response.content.strip() if hasattr(response, 'content') else str(response).strip()
        else:
            answer = str(llm(prompt)).strip()

        return answer
````

## Recommended Compatible LangChain Version Sets

To avoid `ResolutionImpossible` errors, it is best to use one of the following sets. Set A is the most modern, while Set B is highly stable for production-like environments.

### Set A (Modern/Latest)
This set uses the most recent architecture where logic is split into specific packages.
*   **langchain:** `0.3.x` (or leave unpinned)
*   **langchain-core:** `0.3.x`
*   **langchain-community:** `0.3.x`
*   **langsmith:** `0.1.x` or `0.2.x`
*   **langgraph:** `0.2.x`

### Set B (Stable 0.2 series)
Many existing tutorials use this version range.
*   **langchain:** `0.2.11`
*   **langchain-core:** `0.2.23`
*   **langchain-community:** `0.2.10`
*   **langsmith:** `0.1.112`
*   **langgraph-prebuilt:** `0.1.4`

### How to Install
If you want to try the modern set (Set A), use this command:
```bash
!pip install -U langchain langchain-community langchain-core langsmith langgraph
```

If you need to prioritize the specific `langsmith` version from your previous request, we should allow the other packages to find their matching versions automatically rather than forcing incompatible ones.

## Choosing Between Set A and Set B

To help you decide which version set to install, consider the following factors:

### Choose Set A (Modern/Latest) if:
*   **You want the newest features:** LangChain 0.3.x includes the latest improvements, performance optimizations, and integrations with new LLM models.
*   **You are starting a new project:** It is generally better to start with the latest stable version to ensure longevity and receive the latest security updates.
*   **You are using LangGraph:** The newest versions of LangGraph are designed to work seamlessly with the 0.3.x ecosystem.
*   **Note:** Some older tutorials might require minor code adjustments to match the updated API structure.

### Choose Set B (Stable 0.2 series) if:
*   **You are following an existing tutorial:** If your learning material or base code was written in early to mid-2024, it likely targets the 0.2 series.
*   **You need high stability:** This version set is very mature and has had many edge cases resolved, making it reliable for projects that don't need the absolute newest features.
*   **You have existing dependencies:** If your project relies on other libraries that haven't yet updated their compatibility for LangChain 0.3, Set B is the safer choice.

### Recommendation
For most users in Google Colab, **Set A** is recommended. It is forward-looking and `pip` is generally better at resolving dependencies for the most recent versions automatically.

**To install Set A, use:**
```bash
!pip install -U langchain langchain-community langchain-core langsmith langgraph
```

````python
from langchain_huggingface import HuggingFacePipeline
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
import torch

model_id = "Qwen/Qwen2.5-0.5B-Instruct"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype="auto",
    device_map="auto"
)

# Create a transformers pipeline
hf_pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=512,
    temperature=0.7,
    do_sample=True,
    top_p=0.95,
)

# Wrap in LangChain's HuggingFacePipeline
llm = HuggingFacePipeline(pipeline=hf_pipe)

print(f"LLM '{model_id}' initialized via Hugging Face Hub.")

# Simple verification test
try:
    response = llm.invoke("Hello! Tell me a very short joke.")
    print(f"\nVerification Response:\n{response}")
except Exception as e:
    print(f"Verification failed: {e}")
````

## Initialize Qdrant Client

### Subtask:
Ensure the Qdrant client is initialized by running cell `cCRwpKFqDTn9`. This cell contains the code to set up the in-memory Qdrant client and create the 'test' collection.


**Reasoning**:
The subtask requires initializing the Qdrant client and creating a 'test' collection. I will generate a code block with the necessary imports and code to perform these actions.



````python
from qdrant_client import QdrantClient, models
from sentence_transformers import SentenceTransformer
from langchain_qdrant import Qdrant

# Initialize in-memory Qdrant client
client = QdrantClient(':memory:')

# Initialize the embedding model
# This model is consistent with 'thenlper/gte-small' used in previous cells
embeddings = SentenceTransformer('thenlper/gte-small')

# Create the 'test' collection if it doesn't exist
collection_name = 'test'
if not client.collection_exists(collection_name=collection_name):
    client.create_collection(
        collection_name=collection_name,
        vectors_config=models.VectorParams(size=embeddings.get_sentence_embedding_dimension(), distance=models.Distance.COSINE)
    )
    print(f"Collection '{collection_name}' created successfully.")
else:
    print(f"Collection '{collection_name}' already exists.")

print("Qdrant client initialized and 'test' collection ensured.")
````

````python
from transformers import Pipeline, CLIPTextModel, CLIPTokenizer
from diffusers import StableDiffusionPipeline
import torch # Often needed for StableDiffusion and pipelines

class MyPipeline(Pipeline):
    def __init__(self, *args, **kwargs):
        # Determine device: Use CUDA if available, otherwise CPU
        device = "cuda" if torch.cuda.is_available() else "cpu"
        model_id_sd = "runwayml/stable-diffusion-v1-5"

        # Load a compatible CLIPTextModel and CLIPTokenizer to pass to super().__init__
        try:
            text_encoder = CLIPTextModel.from_pretrained(model_id_sd, subfolder="text_encoder")
            tokenizer = CLIPTokenizer.from_pretrained(model_id_sd, subfolder="tokenizer")
        except Exception as e:
            print(f"Warning: Failed to load text_encoder/tokenizer from SD subfolder: {e}. Falling back to generic CLIP.")
            text_encoder = CLIPTextModel.from_pretrained("openai/clip-vit-base-patch32")
            tokenizer = CLIPTokenizer.from_pretrained("openai/clip-vit-base-patch32")

        # Initialize the base Pipeline with the loaded model and tokenizer
        super().__init__(model=text_encoder, tokenizer=tokenizer, task="text-to-image", *args, **kwargs, device=device)

        # Now load the actual StableDiffusionPipeline internally for the text-to-image generation
        if device == "cuda":
            self.pipe = StableDiffusionPipeline.from_pretrained(model_id_sd, torch_dtype=torch.float16)
        else:
            self.pipe = StableDiffusionPipeline.from_pretrained(model_id_sd)
        self.pipe.to(device) # Ensure the pipeline is on the correct device
        self.device = device # Store device for potential use in other methods

    def _sanitize_parameters(self, *args, **kwargs):
        # This method is called internally by Pipeline.__init__ and when the pipeline is called.
        # When called by Pipeline.__init__ for internal checks, args and kwargs might be empty.
        # We need to gracefully handle this case without raising an error.

        # If args is empty and 'prompt' is not in kwargs, it's likely an internal call during init.
        # Return empty lists/dictionaries to satisfy the base class without error.
        if not args and "prompt" not in kwargs:
            return [], {} # Return empty model_inputs, preprocess_inputs

        # For actual user inputs, extract the prompt and prepare for preprocess.
        if args and isinstance(args[0], str):
            # Model inputs are empty, preprocess inputs get the prompt
            return [], {"prompt": args[0], **kwargs}
        elif "prompt" in kwargs:
            # Model inputs are empty, preprocess inputs get the much_needed prompt
            return [], kwargs
        else:
            # This case means an actual call to the pipeline was made with invalid arguments.
            raise ValueError("Input must be a string or include 'prompt' in kwargs")


    def preprocess(self, inputs):
        # During Pipeline.__init__ checks, inputs here could be an empty dictionary (from _sanitize_parameters).
        # The base Pipeline class's `_ensure_processing_supported` method calls `self.model` (our CLIPTextModel)
        # with the output of `preprocess`. `CLIPTextModel` expects tokenized inputs (input_ids), not a raw prompt string.
        # This creates a conflict, as `_forward` needs a prompt string for `StableDiffusionPipeline`.
        # To get past the `Pipeline.__init__` checks for `self.model` (CLIPTextModel), `preprocess` must return something
        # that `CLIPTextModel` can consume. So, if inputs are empty, we return dummy tokenized input.

        # If inputs are empty (from _sanitize_parameters during init), tokenize an empty string for CLIPTextModel.
        if not inputs and not (isinstance(inputs, (str, dict)) and (inputs or "prompt" in inputs)):
            return self.tokenizer("", return_tensors="pt") # Return tokenized empty string for CLIPTextModel compatibility during init

        # For actual calls, inputs should contain the prompt string.
        prompt = None
        if isinstance(inputs, tuple) and inputs and isinstance(inputs[0], str):
            prompt = inputs[0]
        elif isinstance(inputs, dict) and "prompt" in inputs:
            prompt = inputs["prompt"]
        elif isinstance(inputs, str):
            prompt = inputs

        if prompt is None:
            raise ValueError("Input to preprocess must be a string or a dictionary with a 'prompt' key.")

        # For actual pipeline calls, return the prompt as a dict for StableDiffusion
        return {"prompt": prompt}

    def _forward(self, inputs):
        # During Pipeline.__init__ checks, inputs might be tokenized tensors from `preprocess`.
        # During actual calls, inputs should be `{"prompt": "..."}` for `StableDiffusionPipeline`.

        # If inputs are tokenized tensors (e.g., `input_ids`), it's likely from the `Pipeline.__init__` check path
        # for `self.model` (CLIPTextModel). In this case, we just run CLIPTextModel and return its output
        # to satisfy the internal check, without involving `StableDiffusionPipeline`.
        if isinstance(inputs, dict) and "input_ids" in inputs:
            # This path is for the base Pipeline's internal consistency check with `self.model` (CLIPTextModel).
            # We must return an output format that `_postprocess` is prepared to handle as non-image.
            return self.model(**inputs) # Run the CLIPTextModel directly

        # This is the path for actual StableDiffusion generation.
        # Here, `inputs` should be `{"prompt": "..."}` for `self.pipe`.
        return self.pipe(**inputs).images[0]

    def postprocess(self, outputs):
        # During Pipeline.__init__ checks, outputs might be from CLIPTextModel (a tuple of tensors).
        # For actual StableDiffusion calls, outputs are images.

        # If outputs are from CLIPTextModel (e.g., a tuple of tensors), it's from the `Pipeline.__init__` check path.
        # We should not treat this as an image; return None or a dummy value to avoid further processing errors.
        if isinstance(outputs, tuple) and len(outputs) == 1 and isinstance(outputs[0], torch.Tensor):
            # This is the dummy output from CLIPTextModel. Just return it without further image processing.
            return outputs # Or None, if not interested in this intermediate output

        # This is the path for actual StableDiffusion image outputs.
        return outputs

try:
    print("Attempting to instantiate MyPipeline with 'runwayml/stable-diffusion-v1-5'...")
    my_pipeline = MyPipeline()
    print("MyPipeline instantiated successfully.")

    print("\nAttempting to run MyPipeline...")
    result = my_pipeline("A futuristic car driving on a desert road, cinematic, 8k")
    print("MyPipeline executed successfully. Result (first 50 chars of string representation):")
    print(str(result)[:50] + "...")

except Exception as e:
    print(f"Error instantiating or running MyPipeline: {e}")
````

To fix this, I will refine the `_sanitize_parameters`, `preprocess`, and `_forward` methods:
1.  `_sanitize_parameters`: Will extract the prompt and place it directly into `forward_kwargs` for `StableDiffusionPipeline`.
2.  `preprocess`: Will be simplified to always return tokenized empty input, as its primary role in this setup is to satisfy `transformers.Pipeline`'s internal consistency checks for `self.model` (the `CLIPTextModel`).

If the prompt is not in `forward_params`, it will assume it's an internal `CLIPTextModel` check and use `preprocess_outputs` accordingly.
This refactoring ensures `StableDiffusionPipeline` receives its prompt correctly structured and separates the handling of internal `transformers.Pipeline` checks from actual generation logic.



````python
from transformers import Pipeline, CLIPTextModel, CLIPTokenizer
from diffusers import StableDiffusionPipeline
import torch # Often needed for StableDiffusion and pipelines

class MyPipeline(Pipeline):
    def __init__(self, *args, **kwargs):
        # Determine device: Use CUDA if available, otherwise CPU
        device = "cuda" if torch.cuda.is_available() else "cpu"
        model_id_sd = "runwayml/stable-diffusion-v1-5"

        # Load a compatible CLIPTextModel and CLIPTokenizer to pass to super().__init__
        # These will satisfy the transformers.Pipeline base class requirements.
        try:
            text_encoder = CLIPTextModel.from_pretrained(model_id_sd, subfolder="text_encoder")
            tokenizer = CLIPTokenizer.from_pretrained(model_id_sd, subfolder="tokenizer")
        except Exception as e:
            print(f"Warning: Failed to load text_encoder/tokenizer from SD subfolder: {e}. Falling back to generic CLIP.")
            text_encoder = CLIPTextModel.from_pretrained("openai/clip-vit-base-patch32")
            tokenizer = CLIPTokenizer.from_pretrained("openai/clip-vit-base-patch32")

        # Initialize the base Pipeline with the loaded model and tokenizer
        super().__init__(model=text_encoder, tokenizer=tokenizer, task="text-to-image", *args, **kwargs, device=device)

        # Now load the actual StableDiffusionPipeline internally for the text-to-image generation
        if device == "cuda":
            self.pipe = StableDiffusionPipeline.from_pretrained(model_id_sd, torch_dtype=torch.float16)
        else:
            self.pipe = StableDiffusionPipeline.from_pretrained(model_id_sd, torch_dtype=torch.float32) # Use float32 for CPU
        self.pipe.to(device) # Ensure the pipeline is on the correct device
        self.device = device # Store device for potential use in other methods

    def _sanitize_parameters(self, *args, **kwargs):
        model_kwargs = {}
        preprocess_kwargs = {} # Will only be used for internal base Pipeline checks
        forward_kwargs = kwargs.copy()

        # Case 1: Internal Pipeline.__init__ calls without explicit prompts.
        if not args and not kwargs:
            return {}, {}, {}

        # Extract prompt.
        prompt_text = None
        if len(args) == 1 and isinstance(args[0], str):
            prompt_text = args[0]
        elif "prompt" in forward_kwargs: # 'prompt' provided as kwarg
            prompt_text = forward_kwargs.pop("prompt")
        else:
            raise ValueError("Input must be a string or include 'prompt' in kwargs")

        # For actual SD calls, the prompt goes into forward_kwargs.
        # For base Pipeline's internal checks, preprocess_kwargs can be empty.
        forward_kwargs["prompt"] = prompt_text # Add prompt to forward_kwargs for StableDiffusion

        return model_kwargs, preprocess_kwargs, forward_kwargs

    def preprocess(self, inputs):
        # This method is primarily used by the base Pipeline for internal checks (e.g., during __init__ validation).
        # It needs to return something that self.model (CLIPTextModel) can process.
        # For actual user calls, the prompt is already in forward_kwargs and bypasses this for _forward.

        # Always return tokenized empty string for internal base Pipeline checks
        # This ensures self.model (CLIPTextModel) receives a valid input when checked by the base Pipeline.
        return self.tokenizer("", return_tensors="pt")

    def _forward(self, model_inputs, preprocess_outputs, forward_params):
        # model_inputs: {}
        # preprocess_outputs: tokenized empty string from preprocess (used for base Pipeline internal checks with self.model)
        # forward_params: contains the 'prompt' for StableDiffusion, plus any other SD kwargs

        # If 'prompt' is present in forward_params, it's an actual user call for Stable Diffusion.
        if "prompt" in forward_params and isinstance(forward_params["prompt"], str):
            prompt_for_sd = forward_params.pop("prompt") # Get prompt and remove it from forward_params
            return self.pipe(prompt=prompt_for_sd, **forward_params).images[0]

        # Otherwise, it's likely an internal base Pipeline check, which runs self.model (CLIPTextModel).
        # In this case, preprocess_outputs (tokenized empty string) should be fed to self.model.
        if isinstance(preprocess_outputs, dict) and "input_ids" in preprocess_outputs:
            return self.model(**preprocess_outputs)

        # Fallback for unexpected scenarios
        raise RuntimeError("Unexpected _forward call without prompt for StableDiffusion or valid internal check inputs.")


    def postprocess(self, outputs):
        # If outputs are from CLIPTextModel (a tuple of tensors), it's from the `Pipeline.__init__` check path.
        if isinstance(outputs, tuple) and len(outputs) > 0 and isinstance(outputs[0], torch.Tensor):
            return outputs

        # This is the path for actual StableDiffusion image outputs.
        return outputs

try:
    print("Attempting to instantiate MyPipeline with 'runwayml/stable-diffusion-v1-5'...")
    my_pipeline = MyPipeline()
    print("MyPipeline instantiated successfully.")

    print("\nAttempting to run MyPipeline...")
    result = my_pipeline("A futuristic car driving on a desert road, cinematic, 8k")
    print("MyPipeline executed successfully. Result (first 50 chars of string representation):")
    print(str(result)[:50] + "...")

except Exception as e:
    print(f"Error instantiating or running MyPipeline: {e}")
````

````python
from transformers import pipeline
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

READER_MODEL_NAME = "HuggingFaceH4/zephyr-7b-beta"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
)
model = AutoModelForCausalLM.from_pretrained(
    READER_MODEL_NAME, quantization_config=bnb_config
)
tokenizer = AutoTokenizer.from_pretrained(READER_MODEL_NAME)

READER_LLM = pipeline(
    model=model,
    tokenizer=tokenizer,
    task="text-generation",
    do_sample=True,
    temperature=0.2,
    repetition_penalty=1.1,
    return_full_text=False,
    max_new_tokens=500,
)
````

````python
from sentence_transformers import SentenceTransformer

# 1. Initialize the embedding model (384 dimensions)
embed_model = SentenceTransformer('thenlper/gte-small')

def retrieve_context(state: AgentState):
    """Node function to retrieve relevant context from Qdrant."""
    # Get the last message content as the search query
    user_query = state['messages'][-1].content

    # Generate query embedding
    query_vector = embed_model.encode(user_query).tolist()

    # Search the Qdrant collection using the updated API syntax
    search_results = client.query_points(
        collection_name=COLLECTION_NAME,
        query=query_vector,
        limit=3
    )

    # Format the retrieved context into a single string
    # In query_points, the results are in the .points attribute
    context_str = "\n\n".join([hit.payload.get('text', '') for hit in search_results.points])

    print(f"Retrieved {len(search_results.points)} relevant document chunks.")
    return {"context": context_str}

print("Retrieval node logic updated to use Qdrant query_points.")
````

# Task
Build a RAG-based AI agent by first processing and ingesting documentation from the "m-ric/huggingface_doc" dataset into the 'huggingface_docs' Qdrant collection using 'gte-small' embeddings. Then, update the LangGraph workflow to include a 'retrieve' node for context injection. Finally, perform an end-to-end verification to ensure the agent uses the retrieved knowledge to answer queries accurately.

## Process and Ingest Documents

### Subtask:
Load, chunk, embed, and ingest documentation into the Qdrant vector database.


**Reasoning**:
I will load the dataset, process it into chunks using a markdown-aware text splitter, generate embeddings, and then ingest the resulting data points into the Qdrant vector database as specified.



````python
import datasets
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document as LangchainDocument
from sentence_transformers import SentenceTransformer
from qdrant_client import models
from tqdm.auto import tqdm

# 1. Load the dataset
ds = datasets.load_dataset('m-ric/huggingface_doc', split='train')

# 2. Configure the text splitter
MARKDOWN_SEPARATORS = [
    "\n#{1,6} ", "```\n", "\n\\*\\*\\*+\n", "\n---+\n", "\n___+\n", "\n\n", "\n", " ", ""
]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=512,
    chunk_overlap=50,
    add_start_index=True,
    strip_whitespace=True,
    separators=MARKDOWN_SEPARATORS
)

# 3. Process documents and generate chunks
raw_docs = [
    LangchainDocument(page_content=doc['text'], metadata={'source': doc['source']})
    for doc in tqdm(ds, desc='Preparing Documents')
]

docs_processed = text_splitter.split_documents(raw_docs)
print(f'Generated {len(docs_processed)} chunks.')

# 4. Initialize embedding model and generate embeddings
embed_model = SentenceTransformer('thenlper/gte-small')

# 5 & 6. Prepare PointStructs and Ingest into Qdrant
batch_size = 100
for i in tqdm(range(0, len(docs_processed), batch_size), desc='Ingesting to Qdrant'):
    batch = docs_processed[i : i + batch_size]
    texts = [doc.page_content for doc in batch]
    embeddings_list = embed_model.encode(texts).tolist()

    points = [
        models.PointStruct(
            id=i + idx,
            vector=embeddings_list[idx],
            payload={'text': batch[idx].page_content, 'metadata': batch[idx].metadata}
        )
        for idx in range(len(batch))
    ]

    client.upsert(
        collection_name=COLLECTION_NAME,
        points=points
    )

# 7. Verify ingestion
collection_info = client.get_collection(collection_name=COLLECTION_NAME)
print(f'Ingestion complete. Total points in collection: {collection_info.points_count}')
````

````python
from qdrant_client import QdrantClient, models

# Инициализация клиента в оперативной памяти
client = QdrantClient(':memory:')

# Определение имени коллекции
COLLECTION_NAME = 'huggingface_docs'

# Создание коллекции с параметрами векторов для модели gte-small (384 измерения)
if not client.collection_exists(COLLECTION_NAME):
    client.create_collection(
        collection_name=COLLECTION_NAME,
        vectors_config=models.VectorParams(size=384, distance=models.Distance.COSINE)
    )
    print(f"Коллекция '{COLLECTION_NAME}' создана.")
else:
    print(f"Коллекция '{COLLECTION_NAME}' уже существует.")

print("Клиент Qdrant готов к работе.")
````

## Process and Ingest Documents

### Subtask:
Load, chunk, embed, and ingest documentation into the Qdrant vector database.


**Reasoning**:
I will load the dataset, process it into chunks using a markdown-aware text splitter, generate embeddings using the gte-small model, and then ingest the resulting data points into the Qdrant vector database in batches as specified.



````python
import datasets
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document as LangchainDocument
from sentence_transformers import SentenceTransformer
from qdrant_client import models
from tqdm.auto import tqdm

# 1. Load the dataset
ds = datasets.load_dataset('m-ric/huggingface_doc', split='train')

# 2. Configure the text splitter
MARKDOWN_SEPARATORS = [
    "\n#{1,6} ", "```\n", "\n\\*\\*\\*+\n", "\n---+\n", "\n___+\n", "\n\n", "\n", " ", ""
]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=512,
    chunk_overlap=50,
    add_start_index=True,
    strip_whitespace=True,
    separators=MARKDOWN_SEPARATORS
)

# 3. Process documents and generate chunks
raw_docs = [
    LangchainDocument(page_content=doc['text'], metadata={'source': doc['source']})
    for doc in tqdm(ds, desc='Preparing Documents')
]

docs_processed = text_splitter.split_documents(raw_docs)
print(f'Generated {len(docs_processed)} chunks.')

# 4. Initialize embedding model
embed_model = SentenceTransformer('thenlper/gte-small')

# 5, 6 & 7. Prepare PointStructs and Ingest into Qdrant in batches
batch_size = 100
for i in tqdm(range(0, len(docs_processed), batch_size), desc='Ingesting to Qdrant'):
    batch = docs_processed[i : i + batch_size]
    texts = [doc.page_content for doc in batch]
    embeddings_list = embed_model.encode(texts).tolist()

    points = [
        models.PointStruct(
            id=i + idx,
            vector=embeddings_list[idx],
            payload={'text': batch[idx].page_content, 'source': batch[idx].metadata['source']}
        )
        for idx in range(len(batch))
    ]

    client.upsert(
        collection_name=COLLECTION_NAME,
        points=points
    )

# 8. Verify ingestion
collection_info = client.get_collection(collection_name=COLLECTION_NAME)
print(f'Ingestion complete. Total points in collection: {collection_info.points_count}')
````

````python
import datasets
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from sentence_transformers import SentenceTransformer
from qdrant_client import models
from tqdm.auto import tqdm

# 1 & 2. Load the dataset
ds = datasets.load_dataset('m-ric/huggingface_doc', split='train')

# 3. Configure the text splitter with hierarchical Markdown separators
MARKDOWN_SEPARATORS = [
    "\n#{1,6} ", "```\n", "\n\\*\\*\\*+\n", "\n---+\n", "\n___+\n", "\n\n", "\n", " ", ""
]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=512,
    chunk_overlap=50,
    add_start_index=True,
    strip_whitespace=True,
    separators=MARKDOWN_SEPARATORS
)

# 4 & 5. Convert to Document objects and split into chunks
raw_docs = [
    Document(page_content=doc['text'], metadata={'source': doc['source']})
    for doc in tqdm(ds, desc='Preparing Documents')
]

docs_processed = text_splitter.split_documents(raw_docs)
print(f'Generated {len(docs_processed)} chunks.')

# 6. Initialize embedding model
embed_model = SentenceTransformer('thenlper/gte-small')

# 7. Prepare PointStructs and Ingest into Qdrant in batches
batch_size = 100
for i in tqdm(range(0, len(docs_processed), batch_size), desc='Ingesting to Qdrant'):
    batch = docs_processed[i : i + batch_size]
    texts = [doc.page_content for doc in batch]
    embeddings_list = embed_model.encode(texts).tolist()

    points = [
        models.PointStruct(
            id=i + idx,
            vector=embeddings_list[idx],
            payload={
                'text': batch[idx].page_content,
                'source': batch[idx].metadata['source']
            }
        )
        for idx in range(len(batch))
    ]

    client.upsert(
        collection_name=COLLECTION_NAME,
        points=points
    )

# 8. Verify ingestion
collection_info = client.get_collection(collection_name=COLLECTION_NAME)
print(f'Ingestion complete. Total points in collection: {collection_info.points_count}')
````
